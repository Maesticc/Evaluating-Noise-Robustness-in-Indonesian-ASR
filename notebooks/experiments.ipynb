# Small-scale experiments
Listen to pilot mixtures and inspect the achieved SNR.

```python
import os, sys
os.chdir(".."); sys.path.insert(0, ".")
import pandas as pd
from IPython.display import Audio, display
from src.utils import load_config, manifest_path
from src.noise_mixing import generate_waveform
cfg = load_config()
cond = pd.read_csv(manifest_path(cfg, "conditions_test_pilot.csv"))
cond.groupby("condition")[["achieved_snr_db", "output_gain"]].describe()
```

```python
row = cond[cond["utt_id"] == cond["utt_id"].iloc[0]]
for _, r in row.iterrows():
    print(r["condition"], r["noise_category"])
    display(Audio(generate_waveform(cfg, r), rate=16000))
```

```python
# Plan 11.4: inspect waveforms, peak amplitude, clipping, duration and achieved SNR per condition
import matplotlib.pyplot as plt
import numpy as np
fig, axes = plt.subplots(len(row), 1, figsize=(10, 1.6 * len(row)), sharex=True, sharey=True)
for ax, (_, r) in zip(axes, row.iterrows()):
    wav = generate_waveform(cfg, r)
    ax.plot(np.arange(wav.size) / 16000, wav, lw=0.4)
    ax.set_title(f"{r['condition']}  peak={np.abs(wav).max():.3f}  achieved SNR={r['achieved_snr_db']}", fontsize=9)
axes[-1].set_xlabel("time (s)")
plt.tight_layout(); plt.show()
cond[["duration_sec", "peak_before_gain", "output_gain", "clipping_detected", "residual_clipping"]].describe(include="all")
```
